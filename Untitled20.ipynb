```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3
```

```python
df = pd.read_excel(r"C:\Users\Arth Shukhla\Downloads\Dataset2.xls")
```

```python
conn = sqlite3.connect("district_population.db")
df.to_sql("population", conn, if_exists="replace", index=False)
```

```python
df.head()
```

```python
df['Population'] = df['Population'].replace({',': ''}, regex=True).astype(int)
```

```python
query = """
SELECT State, 
       SUM(Population) AS total_population 
FROM population 
GROUP BY State 
ORDER BY total_population DESC
"""
summary_df = pd.read_sql_query(query, conn)
```

```python
print("\nTop 10 States by Population:")
print(summary_df.head(10))
```

```python
summary_df.head(10).plot(kind='bar', x='State', y='total_population', legend=False)
plt.title("Top 10 Most Populous States in India")
plt.xlabel("State")
plt.ylabel("Total Population")
plt.tight_layout()
plt.savefig("population_chart.png")
plt.show()
```
